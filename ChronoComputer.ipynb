````python
# @title
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from dataclasses import dataclass
from typing import Tuple, List, Dict

plt.style.use('dark_background')

# ====================== CONFIGURATION DATACLASS ======================

@dataclass
class SimulationConfig:
    """Centralized configuration for all simulations."""
    GRID_SIZE: int = 100
    STEPS: int = 600
    DT: float = 0.005
    KAPPA: float = 1.0
    SATURATION_THRESHOLD: float = 0.0497
    OPTIMAL_HARVEST_TAU: float = 1.600
    TEMPORAL_WINDOW_SPREAD: float = 0.1
    TEMPORAL_WEIGHT: float = 0.75
    SPATIAL_WEIGHT: float = 0.25
    BOUNDARY_THICKNESS: int = 2

# ====================== HELPER FUNCTIONS ======================

def apply_boundary_conditions(field: np.ndarray, boundary_mask: np.ndarray, value: float = 0.0) -> None:
    """
    Apply boundary conditions to a field.

    Args:
        field: The field array to modify
        boundary_mask: Boolean mask indicating boundary regions
        value: Value to set at boundaries (default 0.0)
    """
    field[boundary_mask] = value

def create_boundary_mask(shape: Tuple[int, ...], thickness: int = 2) -> np.ndarray:
    """
    Create a boundary mask for n-dimensional arrays.

    Args:
        shape: Shape of the array
        thickness: Thickness of boundary in grid points

    Returns:
        Boolean mask with True at boundaries
    """
    mask = np.zeros(shape, dtype=bool)
    for dim in range(len(shape)):
        slices_start = [slice(None)] * len(shape)
        slices_end = [slice(None)] * len(shape)
        slices_start[dim] = slice(0, thickness)
        slices_end[dim] = slice(-thickness, None)
        mask[tuple(slices_start)] = True
        mask[tuple(slices_end)] = True
    return mask

def calculate_legibility(psi: np.ndarray, velocity: np.ndarray, kappa: float,
                        tau: float, config: SimulationConfig) -> float:
    """
    Calculate the Readout Legibility Index.

    Args:
        psi: Displacement field
        velocity: Velocity field
        kappa: Focal curvature parameter
        tau: Current normalized time
        config: Simulation configuration

    Returns:
        Legibility score (0-100)
    """
    laplacian = (np.roll(psi, 1, axis=0) + np.roll(psi, -1, axis=0) +
                 np.roll(psi, 1, axis=1) + np.roll(psi, -1, axis=1) - 4 * psi)

    focal_field = psi - kappa * laplacian * 0.35

    peak_signal = np.max(np.abs(focal_field))
    background_entropy = np.mean(np.abs(focal_field)) + 1e-6
    spatial_sharpness = np.std(focal_field)

    spatial_score = (peak_signal / background_entropy) * spatial_sharpness
    temporal_score = np.max(np.abs(velocity)) / (np.std(psi) + 1e-6)

    legibility = (config.TEMPORAL_WEIGHT * temporal_score +
                  config.SPATIAL_WEIGHT * spatial_score) * 100.0

    return legibility

# ====================== IMPROVED SIMULATIONS ======================

def t0c_second_order_simulation(config: SimulationConfig = None) -> Tuple[List[float], List[float]]:
    """
    Upgraded T'Z0C 2D Second-Order Lattice Simulation.

    Tracks displacement and velocity fields to enable genuine mechanical wave
    propagation, boundary reflections, and true 75/25 temporal-spatial ripening.

    Args:
        config: Simulation configuration object

    Returns:
        Tuple of (tau_values, legibility_history)
    """
    if config is None:
        config = SimulationConfig()

    # Initialize fields
    psi = np.zeros((config.GRID_SIZE, config.GRID_SIZE))
    velocity = np.zeros((config.GRID_SIZE, config.GRID_SIZE))
    theta = np.zeros((config.GRID_SIZE, config.GRID_SIZE))

    # Initialize Gaussian pulse
    x, y = np.meshgrid(np.arange(config.GRID_SIZE), np.arange(config.GRID_SIZE))
    center = config.GRID_SIZE // 2
    r = np.sqrt((x - center)**2 + (y - center)**2)
    psi = np.exp(-r**2 / (2 * 6**2)) * 5.0

    legibility_history = []
    tau_values = []

    boundary_mask = create_boundary_mask(psi.shape, config.BOUNDARY_THICKNESS)

    for t in range(config.STEPS):
        tau = t * config.DT

        # Compute Laplacian
        laplacian = (np.roll(psi, 1, axis=0) + np.roll(psi, -1, axis=0) +
                     np.roll(psi, 1, axis=1) + np.roll(psi, -1, axis=1) - 4 * psi)

        # Update dynamics
        acceleration = 1.2 * laplacian - 0.05 * velocity
        velocity += acceleration * config.DT
        psi += velocity * config.DT

        # Accumulate torsional strain
        theta += 0.08 * np.abs(psi)

        # Apply saturation
        sat_mask = theta > config.SATURATION_THRESHOLD
        velocity[sat_mask] *= -0.80
        theta[sat_mask] *= 0.1

        # Apply boundary conditions
        apply_boundary_conditions(psi, boundary_mask)
        apply_boundary_conditions(velocity, boundary_mask)

        # Sample legibility periodically
        if t % 6 == 0 and tau > 0.2:
            leg = calculate_legibility(psi, velocity, config.KAPPA, tau, config)
            legibility_history.append(leg)
            tau_values.append(tau)

    return tau_values, legibility_history

def execute_synthetic_ratio_sweep(config: SimulationConfig = None) -> Tuple[np.ndarray, List[float], List[float]]:
    """
    Performs a synthetic parameter sweep across temporal/spatial readout weights.

    Args:
        config: Simulation configuration object

    Returns:
        Tuple of (temporal_weights, peak_legibility_scores, optimal_taus)
    """
    if config is None:
        config = SimulationConfig(GRID_SIZE=80, STEPS=400)

    temporal_weights = np.linspace(0.50, 0.95, 46)
    peak_legibility_scores = []
    optimal_taus = []

    # Initialize pulse
    x, y = np.meshgrid(np.arange(config.GRID_SIZE), np.arange(config.GRID_SIZE))
    center = config.GRID_SIZE // 2
    r = np.sqrt((x - center)**2 + (y - center)**2)
    initial_pulse = np.exp(-r**2 / (2 * 5**2)) * 5.0

    boundary_mask = create_boundary_mask(initial_pulse.shape, config.BOUNDARY_THICKNESS)

    for w_temporal in temporal_weights:
        w_spatial = 1.0 - w_temporal

        psi = np.copy(initial_pulse)
        velocity = np.zeros((config.GRID_SIZE, config.GRID_SIZE))
        theta = np.zeros((config.GRID_SIZE, config.GRID_SIZE))

        max_legibility = 0.0
        best_tau = 0.0

        for t in range(config.STEPS):
            tau = t * config.DT

            laplacian = (np.roll(psi, 1, axis=0) + np.roll(psi, -1, axis=0) +
                         np.roll(psi, 1, axis=1) + np.roll(psi, -1, axis=1) - 4 * psi)

            acceleration = 1.2 * laplacian - 0.04 * velocity
            velocity += acceleration * config.DT
            psi += velocity * config.DT

            theta += 0.08 * np.abs(psi)
            sat_mask = theta > config.SATURATION_THRESHOLD
            velocity[sat_mask] *= -0.80
            theta[sat_mask] *= 0.1

            apply_boundary_conditions(psi, boundary_mask)
            apply_boundary_conditions(velocity, boundary_mask)

            if t % 5 == 0 and tau > 0.3:
                focal_field = psi - config.KAPPA * laplacian * 0.35

                peak_signal = np.max(np.abs(focal_field))
                background_entropy = np.mean(np.abs(focal_field)) + 1e-6
                spatial_sharpness = np.std(focal_field)

                spatial_score = (peak_signal / background_entropy) * spatial_sharpness
                temporal_score = np.max(np.abs(velocity)) / (np.std(psi) + 1e-6)

                combined_legibility = (w_temporal * temporal_score) + (w_spatial * spatial_score)

                if combined_legibility > max_legibility:
                    max_legibility = combined_legibility
                    best_tau = tau

        peak_legibility_scores.append(max_legibility)
        optimal_taus.append(best_tau)

    return temporal_weights, peak_legibility_scores, optimal_taus

def t0c_3d_tuned_sim(size=55, steps=1300, dt=0.004, kappa=1.15, tetra_bias=0.42):
    """Tuned 3D version with realistic normalization."""
    psi = np.zeros((size, size, size))
    vel = np.zeros((size, size, size))
    theta = np.zeros((size, size, size))

    center = size // 2
    x, y, z = np.mgrid[0:size, 0:size, 0:size]
    r = np.sqrt((x-center)**2 + (y-center)**2 + (z-center)**2)
    psi = 4.5 * np.exp(-r**2 / 32)

    leg_history, tau_history = [], []

    boundary_mask = np.zeros((size,size,size), dtype=bool)
    # Set boundary conditions for 3D simulation
    for dim in range(3):
        slices_start = [slice(None)] * 3
        slices_end = [slice(None)] * 3
        slices_start[dim] = slice(0, 2)
        slices_end[dim] = slice(-2, None)
        boundary_mask[tuple(slices_start)] = True
        boundary_mask[tuple(slices_end)] = True

    for t in range(steps):
        tau = t * dt
        lap = (np.roll(psi,1,0) + np.roll(psi,-1,0) +
               np.roll(psi,1,1) + np.roll(psi,-1,1) +
               np.roll(psi,1,2) + np.roll(psi,-1,2) - 6*psi)

        bias = tetra_bias * (np.roll(psi,2,0) + np.roll(psi,-2,0) - 2*psi)

        accel = 1.08 * (lap + bias) - 0.038 * vel
        vel += accel * dt
        psi += vel * dt

        theta += 0.055 * np.abs(psi)
        sat = theta > 0.0497
        vel[sat] *= -0.75
        theta[sat] *= 0.22

        # Apply 3D boundaries
        psi[boundary_mask] = 0.0
        vel[boundary_mask] = 0.0

        if t % 7 == 0 and tau > 0.7:
            focal = psi.mean(axis=2) - kappa * lap.mean(axis=2) * 0.4  # Project to 2D for readout

            peak = np.max(np.abs(focal))
            bg = np.mean(np.abs(focal)) + 1e-6
            sharp = np.std(focal)

            spatial_score = (peak / bg) * sharp * 0.8   # Normalized
            temporal_score = np.max(np.abs(vel.mean(axis=2))) / (np.std(psi.mean(axis=2)) + 1e-6)

            leg = (0.75 * temporal_score + 0.25 * spatial_score) * 100
            leg_history.append(leg)
            tau_history.append(tau)

    return tau_history, leg_history

def execute_refraction_path_sweep(grid_size=120, steps=500, dt=0.02):
    """
    Simulates wave propagation across a dual-velocity boundary (Refraction).
    Sweeps multiple trajectory angles to prove that the path minimizing total
    travel time yields the earliest, highest-contrast legibility handshake (τ).
    """
    # Define trajectory angles to test (in degrees relative to vertical)
    test_angles = np.linspace(-30, 30, 7)
    arrival_times = []
    peak_legibilities = []

    # Define source and target positions
    source_x, source_y = grid_size // 4, grid_size // 2
    target_x, target_y = 3 * grid_size // 4, grid_size // 2

    # Create a dual-velocity medium (Top half is fast, bottom half is slow)
    velocity_map = np.ones((grid_size, grid_size)) * 20.0  # Increased Fast zone velocity
    velocity_map[grid_size // 2:, :] = 5.0               # Decreased Slow zone velocity for higher contrast

    # Hard boundary mask to prevent edge bleeding
    boundary_mask = np.zeros((grid_size, grid_size), dtype=bool)
    boundary_mask[0:2, :] = True; boundary_mask[-2:, :] = True
    boundary_mask[:, 0:2] = True; boundary_mask[:, -2:] = True

    # Threshold for a 'significant' legibility to consider it a handshake event
    handshake_threshold = 0.001 # Even lower to ensure detection

    for angle in test_angles:
        # Reset mechanical fields for each trajectory test
        psi = np.zeros((grid_size, grid_size))
        velocity = np.zeros((grid_size, grid_size))
        theta = np.zeros((grid_size, grid_size))

        # Inject an initial directional pulse packet biased along the current angle
        rad = np.radians(angle)
        x, y = np.meshgrid(np.arange(grid_size), np.arange(grid_size))
        r = np.sqrt((x - source_x)**2 + (y - source_y)**2)

        # Directional wave packet initialization (Increased amplitude and sharper pulse)
        directional_bias = np.cos(rad) * (x - source_x) + np.sin(rad) * (y - source_y)
        psi = np.exp(-r**2 / (2 * 2**2)) * np.cos(2 * np.pi * directional_bias / 8.0) * 50.0 # Sharper pulse, increased amplitude

        optimal_tau = dt * steps # Default to max time if no significant legibility
        max_legibility_observed = 0.0

        for t in range(steps):
            tau = t * dt

            # Discrete Laplacian
            laplacian = (np.roll(psi, 1, axis=0) + np.roll(psi, -1, axis=0) +
                         np.roll(psi, 1, axis=1) + np.roll(psi, -1, axis=1) - 4 * psi)

            # Acceleration incorporating the local medium velocity field (further reduced damping)
            acceleration = (velocity_map**2) * laplacian - 0.0001 * velocity # Damping further reduced
            velocity += acceleration * dt
            psi += velocity * dt

            # Accumulate torsional strain and apply saturation mask (adjusted for higher amplitude)
            theta += 0.01 * np.abs(psi) # Adjusted theta accumulation
            sat_mask = theta > 0.0497
            velocity[sat_mask] *= -0.3  # Slightly more reflection
            theta[sat_mask] *= 0.08     # Slightly stronger theta reset

            psi[boundary_mask] = 0.0
            velocity[boundary_mask] = 0.0

            # Monitor a localized readout window surrounding the target node
            if t > 50: # Start monitoring after initial propagation
                readout_zone = psi[target_x-3:target_x+4, target_y-3:target_y+4]
                if readout_zone.size == 0 or np.all(readout_zone == 0):
                    legibility = 0.0 # No signal in zone
                else:
                    spatial_sharpness = np.std(readout_zone)
                    peak_signal = np.max(np.abs(readout_zone))
                    local_entropy = np.mean(np.abs(readout_zone)) + 1e-6
                    legibility = (peak_signal / local_entropy) * spatial_sharpness

                # Capture the *first* time it crosses the handshake threshold
                if legibility > handshake_threshold and optimal_tau == dt * steps:
                    optimal_tau = tau

                # Keep track of the maximum legibility observed for this angle
                if legibility > max_legibility_observed:
                    max_legibility_observed = legibility

        arrival_times.append(optimal_tau) # Store the first time threshold is crossed, or max time
        peak_legibilities.append(max_legibility_observed) # Store the overall peak legibility for plotting

    return test_angles, arrival_times, peak_legibilities

# ====================== EXECUTION & PLOTTING ======================

# Run all simulations

# Simulation 1: T'Z0C Second-Order Simulation (2D)
print("Executing T'Z0C Second-Order Simulation (2D)...")
config_2d = SimulationConfig(TEMPORAL_WEIGHT=0.75, SPATIAL_WEIGHT=0.25) # Use default 75/25 for this simulation
taus_sim1, legibility_scores_sim1 = t0c_second_order_simulation(config_2d)
optimal_idx_sim1 = np.argmax(legibility_scores_sim1)
optimal_tau_sim1 = taus_sim1[optimal_idx_sim1]

# Simulation 2: Synthetic Ratio Sweep
print("Initiating Synthetic Sweep across Temporal/Spatial Ratios...")
config_sweep = SimulationConfig(GRID_SIZE=80, STEPS=400) # Configuration for sweep
weights_sim2, scores_sim2, taus_sim2 = execute_synthetic_ratio_sweep(config_sweep)
best_idx_sim2 = np.argmax(scores_sim2)
opt_temporal_sim2 = weights_sim2[best_idx_sim2] * 100.0
opt_spatial_sim2 = (1.0 - weights_sim2[best_idx_sim2]) * 100.0

# Simulation 3: T'Z0C 3D Tuned Simulation
print("Executing T'Z0C 3D Tuned Simulation...")
taus_sim3, legs_sim3 = t0c_3d_tuned_sim()
peak_idx_sim3 = np.argmax(legs_sim3)

# Simulation 4: Refraction Path Sweep
print("Running Refraction Path Sweep...")
angles_sim4, arrivals_sim4, peak_legibilities_sim4 = execute_refraction_path_sweep()
winning_idx_sim4 = np.argmin(arrivals_sim4)

print("\n--- Simulation Results ---")
print(f"T'Z0C 2D Simulation: Optimal Harvest Window: τ ≈ {optimal_tau_sim1:.3f}, Peak Legibility: {legibility_scores_sim1[optimal_idx_sim1]:.2f}%")
print(f"Synthetic Sweep: Empirical Convergence Peak: {opt_temporal_sim2:.1f}% Temporal / {opt_spatial_sim2:.1f}% Spatial")
print(f"T'Z0C 3D Tuned Simulation: Peak at τ ≈ {taus_sim3[peak_idx_sim3]:.3f} | Legibility ≈ {legs_sim3[peak_idx_sim3]:.1f}%")
print(f"Refraction Path Sweep: Winning Trajectory Angle: {angles_sim4[winning_idx_sim4]:.1f}° relative to vertical axis, Earliest Collapse Handshake: τ ≈ {arrivals_sim4[winning_idx_sim4]:.3f}, Peak Legibility: {peak_legibilities_sim4[winning_idx_sim4]:.2f}")

# 4-Panel Figure
fig = plt.figure(figsize=(22, 13))
gs = fig.add_gridspec(2, 2, hspace=0.3, wspace=0.25)

# Panel 1: 2D Ripening
ax1 = fig.add_subplot(gs[0, 0])
ax1.plot(taus_sim1, legibility_scores_sim1, 'b-', linewidth=2.5, label='Lattice Legibility Trace')
ax1.axvline(x=optimal_tau_sim1, color='red', linestyle='--', linewidth=1.5,
            label=f'Optimal Single-Shot Harvest (τ ≈ {optimal_tau_sim1:.3f})')
ax1.axvline(x=1.600, color='green', alpha=0.6, linestyle=':', linewidth=1.5,
            label='Target Registry Paradigm (τ ≈ 1.600)')
ax1.set_title("T'Z0C Timeline Ripening Window (2D, 75/25 Weighted)", fontsize=12, fontweight='bold')
ax1.set_xlabel("Normalized Time τ (Event Lifecycle Timeline)", fontsize=11)
ax1.set_ylabel("Readout Legibility Matrix (%)", fontsize=11)
ax1.legend(loc='upper right')
ax1.grid(True, alpha=0.25)

# Panel 2: Ratio Sweep
ax2 = fig.add_subplot(gs[0, 1])
ax2.plot(weights_sim2 * 100.0, scores_sim2, 'g-', linewidth=2.5, label='Readout Contrast Signature')
ax2.axvline(x=75.0, color='red', linestyle='--', alpha=0.7,
            label='Theoretical Tetrahedral Lock Reference (75.0 / 25.0)')
ax2.axvline(x=opt_temporal_sim2, color='blue', linestyle=':',
            label=f'Empirical Sweep Peak ({opt_temporal_sim2:.1f} / {opt_spatial_sim2:.1f})')
ax2.set_title("T'Z0C Readout Gate Optimization Sweep", fontsize=12, fontweight='bold')
ax2.set_xlabel("Temporal Pulse Timing Weight (Wt %)", fontsize=11)
ax2.set_ylabel("Unified Readout Legibility Index", fontsize=11)
ax2.legend(loc='lower left')
ax2.grid(True, alpha=0.25)

# Panel 3: 3D Anisotropic (Tuned)
ax3 = fig.add_subplot(gs[1, 0])
ax3.plot(taus_sim3, legs_sim3, 'c-', linewidth=2.5)
ax3.axvline(taus_sim3[peak_idx_sim3], color='red', linestyle='--', label=f'Peak τ = {taus_sim3[peak_idx_sim3]:.3f}')
ax3.axvline(1.600, color='yellow', linestyle=':', label='Target τ = 1.600')
ax3.set_title("3D Tuned Tetrahedral Lattice - Ripening Window", fontsize=12, fontweight='bold')
ax3.set_xlabel("Normalized Time τ", fontsize=11)
ax3.set_ylabel("75/25 Weighted Legibility", fontsize=11)
ax3.legend()
ax3.grid(True, alpha=0.3)

# Panel 4: Refraction Path Sweep
ax4 = fig.add_subplot(gs[1, 1])
ax4.plot(angles_sim4, arrivals_sim4, 'r-o', linewidth=2, markersize=8, label='Time to Handshake Collapse (τ)')
ax4.axvline(x=angles_sim4[winning_idx_sim4], color='blue', linestyle='--', label=f'Optimal Refractive Path ({angles_sim4[winning_idx_sim4]:.1f}°)')
ax4.set_title("T'Z0C Optical Least-Time Path Verification\nCompetitive Wave-Front Collapse Sweep", fontsize=12, fontweight='bold')
ax4.set_xlabel("Initial Launch Trajectory Angle (Degrees)", fontsize=11)
ax4.set_ylabel("Required Handshake Maturity Time (τ)", fontsize=11)
ax4.legend()
ax4.grid(True, alpha=0.25)

# Add overall title / summary
fig.suptitle("T'Z0C Simulation Validation Suite\nTemporal Ripening, Readout Optimization, Anisotropy & Least-Time Routing",
             fontsize=16, fontweight='bold', y=0.98)

# Replace plt.tight_layout() with this for clean multi-panel subplots:
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1, wspace=0.3, hspace=0.4)

plt.show()
````
