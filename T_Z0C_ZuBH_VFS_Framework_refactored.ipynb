## 1. Imports & Framework Constants

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.stats import linregress
import glob, os
import pandas as pd

plt.style.use('seaborn-darkgrid')

# Core geometric / framework constants (compact)
theta_0 = np.arcsin(1/3)                 # diode shear (rad)
p_c     = 5/(32*np.pi)                   # vector saturation
gamma   = p_c * np.sin(theta_0)          # dynamic damping (per-cycle)
Gamma_lattice = 5/(96*np.pi**2)          # framework lattice linewidth (for reference)
sigma = 1.0                               # complementarity scale (dimensionless)

print(f'θ₀ = {theta_0:.6f} rad, γ = {gamma:.8f}, Γ_lattice = {Gamma_lattice:.8f}')
```

## 2. Compact Solvers (discrete & continuous)

```python
def discrete_transition(steps, X0=1.0, Y0=0.0, Phi0=0.0, sigma=1.0):
    X = np.empty(steps+1)
    Y = np.empty(steps+1)
    Phi = np.empty(steps+1)
    X[0], Y[0], Phi[0] = X0, Y0, Phi0
    c, s = np.cos(theta_0), np.sin(theta_0)
    for n in range(steps):
        r2 = X[n]*X[n] + Y[n]*Y[n]
        w = 1.0/(1.0 + r2 / sigma**2)
        X[n+1] = (1-gamma)*w*( X[n]*c - Y[n]*s )
        Y[n+1] = (1-gamma)*w*( X[n]*s + Y[n]*c )
        Phi[n+1] = Phi[n] + gamma*r2
    return X, Y, Phi

def continuous_rhs(t, S, sigma):
    X, Y, Phi = S
    r2 = X*X + Y*Y
    lam = r2/(sigma*sigma + r2)
    dX = -(gamma + lam)*X - theta_0*Y
    dY =  theta_0*X - (gamma + lam)*Y
    dPhi = gamma * r2
    return [dX, dY, dPhi]
```

## 3. Single-run orchestration (compute results used by plotting)
Call this cell to (re)compute the master results. Parameters are local and easy to tune.


```python
# Parameters
cycles = 5000
sigma = 1.0
# Solve continuous system once (dense output for fine plotting)
sol = solve_ivp(continuous_rhs, [0, cycles], [1.0, 0.0, 0.0], args=(sigma,), rtol=1e-9, atol=1e-11, method='RK45', dense_output=True)
if not sol.success:
    raise RuntimeError('continuous solver failed: '+str(sol.message))
t_fine = np.linspace(0, cycles, 5*cycles+1)
S_c_fine = sol.sol(t_fine)
X_c, Y_c, Phi_c = S_c_fine[0], S_c_fine[1], S_c_fine[2]
t_int = np.arange(cycles+1)
S_c_int = sol.sol(t_int)
X_ci, Y_ci, Phi_ci = S_c_int[0], S_c_int[1], S_c_int[2]
# Discrete
X_d, Y_d, Phi_d = discrete_transition(cycles, sigma=sigma)
# Derived quantities
Phi_inf_d = Phi_d[-1]
A_d = X_d**2 + Y_d**2
# Early-time log regression window (automatic but conservative)
N_full = np.arange(1, cycles+1)
lnN_full = np.log(N_full)
Phi_full = Phi_d[1:]
N_cut = min(max(80, 20), 150)
N_win = N_full[:N_cut]
lnN_win = lnN_full[:N_cut]
Phi_win = Phi_full[:N_cut]
slope, intercept, r_reg, *_ = linregress(lnN_win, Phi_win)
print(f'Computed: Φ_∞ (discrete) = {Phi_inf_d:.6f}, early-log slope a = {slope:.6e}, r={r_reg:.4f}')
```

## 4. Multi-panel Figure Suite: Validation + Transient (combined)
This suite combines the discrete-vs-continuous validation and the transient window into a single multi-panel figure for compact review. Only panels that illustrate differences are shown.


```python
def plot_validation_and_transient(t_fine, X_c, Y_c, Phi_c, t_int, X_d, Y_d, Phi_d, transient_cycles=200):
    fig, axes = plt.subplots(2, 3, figsize=(16, 8))
    fig.suptitle('Validation & Transient Suite', fontsize=14)
    # Validation: X, Y, Phi over full range (continuous vs discrete markers)
    axes[0,0].plot(t_fine, X_c, color='C0', label='Cont')
    axes[0,0].scatter(t_int[::max(1,len(t_int)//800)], X_d[::max(1,len(t_int)//800)], color='C3', s=6, label='Disc')
    axes[0,0].set_title('X (Travel)')
    axes[0,0].legend()
    axes[0,1].plot(t_fine, Y_c, color='C1')
    axes[0,1].scatter(t_int[::max(1,len(t_int)//800)], Y_d[::max(1,len(t_int)//800)], color='C4', s=6)
    axes[0,1].set_title('Y (Torque)')
    axes[0,2].plot(t_fine, Phi_c, color='C2')
    axes[0,2].scatter(t_int[::max(1,len(t_int)//800)], Phi_d[::max(1,len(t_int)//800)], color='C5', s=6)
    axes[0,2].set_title('Φ (Phase)')
    # Transient zooms (second row)
    mask_fine = t_fine <= transient_cycles
    mask_int = t_int <= transient_cycles
    axes[1,0].plot(t_fine[mask_fine], X_c[mask_fine], color='C0')
    axes[1,0].scatter(t_int[mask_int], X_d[mask_int], color='C3', s=12)
    axes[1,0].set_title('X transient (N≤{0})'.format(transient_cycles))
    axes[1,1].plot(t_fine[mask_fine], Y_c[mask_fine], color='C1')
    axes[1,1].scatter(t_int[mask_int], Y_d[mask_int], color='C4', s=12)
    axes[1,1].set_title('Y transient')
    axes[1,2].plot(t_fine[mask_fine], Phi_c[mask_fine], color='C2')
    axes[1,2].scatter(t_int[mask_int], Phi_d[mask_int], color='C5', s=12)
    axes[1,2].set_title('Φ transient')
    for axrow in axes:
        for ax in axrow:
            ax.grid(alpha=0.3)
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

plot_validation_and_transient(t_fine, X_c, Y_c, Phi_c, t_int, X_d, Y_d, Phi_d, transient_cycles=200)
```

## 5. Multi-panel Phase & Alpha Suite
Combines phase growth/saturation panels and alpha-function comparisons into a single figure. The alpha functions are compactly defined here so they are easy to test and extend.


```python
def alpha_canonical(Phi):
    C = 137.75937
    return 1.0/(C*(1.0+Phi))
def alpha_exponential(Phi):
    return (1/137.036)*np.exp(-Phi)
def alpha_fabry_perot(Phi):
    return (1/137.036)/(1.0 + Phi + Phi**2)

def plot_phase_and_alpha(Phi_d, A_d, N_full, lnN_full, Phi_full, slope, intercept):
    fig, axes = plt.subplots(2, 2, figsize=(12, 9))
    fig.suptitle('Phase & Alpha Suite', fontsize=14)
    # Φ(N) full
    axes[0,0].plot(np.arange(len(Phi_d)), Phi_d, color='C2')
    axes[0,0].set_title('Φ(N) - Phase Accumulation')
    # Φ vs ln(N) early-window
    axes[0,1].scatter(lnN_full, Phi_full, s=6, color='C2', alpha=0.6)
    axes[0,1].plot(lnN_full[:len(lnN_full)], slope*lnN_full[:len(lnN_full)] + intercept, 'm--', label=f'lin fit (early)')
    axes[0,1].set_title('Φ vs ln(N) - early window')
    axes[0,1].legend()
    # A(N) amplitude (log scale)
    axes[1,0].plot(np.arange(len(A_d)), A_d, color='orange')
    axes[1,0].set_yscale('log')
    axes[1,0].set_title('A(N) - Amplitude Squared')
    # Alpha comparison at selected Phi values (trajectory)
    phis = np.linspace(0, np.max(Phi_d), 200)
    axes[1,1].plot(phis, alpha_canonical(phis), label='canonical')
    axes[1,1].plot(phis, alpha_exponential(phis), label='exponential')
    axes[1,1].plot(phis, alpha_fabry_perot(phis), label='fabry-perot')
    axes[1,1].axvline(Phi_d[-1], color='gray', ls=':', label='Φ_∞')
    axes[1,1].set_title('α(Φ) comparisons')
    axes[1,1].legend()
    for ax in axes.ravel():
        ax.grid(alpha=0.3)
    plt.tight_layout(rect=[0,0.03,1,0.95])
    plt.show()

plot_phase_and_alpha(Phi_d, A_d, N_full, lnN_full, Phi_full, slope, intercept)
```

## 6. Optional: CSV loader + combined multi-file panel
If your project produces CSVs (bench outputs, logs), place them under `data/` and this cell will attempt to load and plot them concisely. If no CSVs are present the cell will print a message and skip.


```python
def load_and_plot_csvs(data_dir='data', max_files=6):
    files = sorted(glob.glob(os.path.join(data_dir, '*.csv')))
    if len(files) == 0:
        print('No CSV files found in', data_dir)
        return
    files = files[:max_files]
    n = len(files)
    fig, axes = plt.subplots(n, 1, figsize=(10, 3*n), sharex=False)
    if n == 1:
        axes = [axes]
    for ax, f in zip(axes, files):
        df = pd.read_csv(f)
        # Heuristics: plot numeric columns only, up to first 5 cols
        numcols = df.select_dtypes(include=[np.number]).columns[:5]
        for col in numcols:
            ax.plot(df.index, df[col], label=col)
        ax.set_title(os.path.basename(f))
        ax.legend(fontsize=8)
        ax.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

# Example: run if you have CSVs
# load_and_plot_csvs('data')
```
