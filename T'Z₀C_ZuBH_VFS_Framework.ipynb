```python
# @title
# ============================================================
# 1. Framework Constants
# ============================================================
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.stats import pearsonr, linregress

plt.style.use("dark_background")

theta_0 = np.arcsin(1/3)                 # diode shear
p_c     = 5/(32*np.pi)                   # vector saturation
gamma   = p_c * np.sin(theta_0)          # = p_c / 3
sigma   = 1.0                            # complementarity scale

print(f"θ₀ = {theta_0:.6f} rad")
print(f"γ  = {gamma:.8f}")
print(f"Γ_lattice (framework) = {5/(96*np.pi**2):.8f}")   # for reference
```

```python
# @title
# ============================================================
# 2. Discrete Solver
# ============================================================
def discrete_transition(steps, X0=1.0, Y0=0.0, Phi0=0.0, sigma=1.0):
    X   = np.empty(steps+1)
    Y   = np.empty(steps+1)
    Phi = np.empty(steps+1)

    X[0], Y[0], Phi[0] = X0, Y0, Phi0

    c, s = np.cos(theta_0), np.sin(theta_0)

    for n in range(steps):
        r2  = X[n]*X[n] + Y[n]*Y[n]
        w   = 1.0 / (1.0 + r2 / sigma**2)
        X[n+1] = (1-gamma)*w*( X[n]*c - Y[n]*s )
        Y[n+1] = (1-gamma)*w*( X[n]*s + Y[n]*c )
        Phi[n+1] = Phi[n] + gamma*r2
    return X, Y, Phi
```

```python
# @title
# ============================================================
# 3. Continuous Solver
# ============================================================
def continuous_rhs(t, S, sigma):
    X, Y, Phi = S
    r2  = X*X + Y*Y
    lam = r2 / (sigma*sigma + r2)
    dX  = -(gamma + lam)*X - theta_0*Y
    dY  =  theta_0*X - (gamma + lam)*Y
    dPhi= gamma * r2
    return [dX, dY, dPhi]
```

```python
# @title
# ============================================================
# 4. Numerical Integration
# ============================================================
cycles = 10000

# Solve the continuous system once with dense_output=True
sol = solve_ivp(continuous_rhs, [0, cycles], [1.0, 0.0, 0.0],
                args=(sigma,), rtol=1e-9, atol=1e-11, method="RK45",
                dense_output=True)
assert sol.success, sol.message

# Evaluate the solution at fine time steps for plotting
t_fine = np.linspace(0, cycles, 5*cycles+1)
S_c_fine = sol.sol(t_fine)
X_c, Y_c, Phi_c = S_c_fine[0], S_c_fine[1], S_c_fine[2]

# Evaluate the solution at integer time steps for comparison with discrete
t_int = np.arange(cycles+1)
S_c_int = sol.sol(t_int)
X_ci, Y_ci, Phi_ci = S_c_int[0], S_c_int[1], S_c_int[2]

# Solve the discrete system
X_d, Y_d, Phi_d = discrete_transition(cycles)

print(f"\nΦ_∞ discrete   = {Phi_d[-1]:.6f}")
print(f"Φ_∞ continuous = {Phi_ci[-1]:.6f}")
print(f"max |X_d − X_c| on integers = {np.max(np.abs(X_d-X_ci)):.4e}")
print(f"max |Y_d − Y_c| on integers = {np.max(np.abs(Y_d-Y_ci)):.4e}")
print(f"max |Φ_d − Φ_c| on integers = {np.max(np.abs(Phi_d-Phi_ci)):.4e}")
```

### Geometric Tail Formulae

For `A_n` once `A_n << σ²`, the decay is approximately geometric.

`rho = (1 - gamma)**2`

`A_tail = A_vals[n0] * (rho ** np.arange(len(A_vals)-n0))`

`Phi_deficit_analytic = (gamma / (1 - rho)) * A_tail`

**Note:** The analytical derivations in this section confirm the numerical convergence properties of the model, specifically demonstrating the exponential decay of the phase deficit and amplitude once the system enters the small-amplitude regime. This provides strong validation for the theoretical understanding of the system's long-term behavior.

```python
# @title
# ============================================================
# 11. Asymptotic Theory (Calculations)
# ============================================================

# Calculate the contraction ratio rho
rho   = (1 - gamma)**2                    # contraction ratio

# Define n0 for starting the asymptotic tail comparison
n0 = 50

# Ensure A_d has enough elements before slicing
if len(A_d) > n0:
    # Calculate analytic amplitude tail (pure geometric from a late but reliable index)
    A_tail = A_d[n0] * (rho ** np.arange(len(A_d)-n0))
    # Calculate analytic phase deficit
    Phi_deficit_analytic = (gamma / (1 - rho)) * A_tail
else:
    # Handle cases where A_d is too short
    A_tail = np.array([])
    Phi_deficit_analytic = np.array([])
    print("Warning: A_d array is too short for asymptotic tail calculation with n0=", n0)


print(f"\ncontraction ratio ρ = (1-γ)² = {rho:.8f}")
print(f"analytic prefactor γ/(1-ρ)   = {gamma/(1-rho):.6f}")
print(f"Φ_∞ (numeric)               = {Phi_inf:.6f}")

print("\nResidual check at a few late points:")
print(f"{'N':<4} {'numeric deficit':<20} {'analytic':<20} {'rel.err':<10}")
for i, n_val in enumerate([100, 200, 500, 1000]):
    if n_val < len(Phi_d) and n_val - n0 < len(Phi_deficit_analytic):
        num = Phi_inf - Phi_d[n_val]
        ana = Phi_deficit_analytic[n_val - n0]
        # Handle potential division by zero for relative error if num is very small
        if num != 0:
            rel_err = abs(num - ana) / num
        else:
            rel_err = np.nan # Or choose a suitable representation for undefined error

        print(f"{n_val:<4d} {num:<20.3e} {ana:<20.3e} {rel_err:<10.2e}")
    elif n_val < len(Phi_d):
        print(f"{n_val:<4d} {'Numeric underflow':<20} {'Analytic not calc.':<20} {'N/A':<10}")
    else:
        print(f"{n_val:<4d} {'N/A':<20} {'N/A':<20} {'N/A':<10}")
```

## Route C: Complementarity Weight / Dual-Ledger Leakage

We will investigate the candidate coupling $\alpha(A) = c_3 w(A)[1-w(A)]$, where $A = X^2 + Y^2$ is the total amplitude and $w(A) = \frac{1}{1 + A/\sigma^2}$ is the weight function. We will then compute its numerical $\beta$-function as $\beta(N) \approx \frac{\alpha(N+1) - \alpha(N)}{\ln(N+1) - \ln N}$ over the discrete simulation cycles.

The constant $c_3$ is currently set to 1.0 as a placeholder, awaiting further investigation to determine its relation to lattice factors.

```python
# @title
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 0. Prerequisites (rely on globals from prior cells)
# ------------------------------------------------------------
# Expected globals from prior cells:
#   X_d, Y_d   : discrete trajectory arrays (length cycles+1)
#   sigma      : complementarity scale (usually 1.0)
#   cycles     : integer, number of map steps
#   N_full     : np.arange(0, cycles+1) or equivalent
#   lnN_full   : np.log(N_full) with a safe floor for N=0

# --- Ensure consistent N_full and lnN_full for this cell ---
# This explicitly re-initializes N_full and lnN_full based on the 'cycles' value
# (from global) to ensure correct lengths, as it might have been modified by previous cells.
N_full   = np.arange(cycles + 1, dtype=float)
# Ensure lnN_full does not have log(0) for N=0
lnN_full = np.log(np.maximum(N_full, 1.0))
# --- End of consistency check ---

# ------------------------------------------------------------
# 1. Core quantities
# ------------------------------------------------------------
A_d = X_d**2 + Y_d**2
w_d = 1.0 / (1.0 + A_d / sigma**2)

def make_alpha(c3, w=w_d):
    """Route-C Form 3: α = c₃ · w · (1 - w)"""
    return c3 * w * (1.0 - w)

def make_beta_std(alpha, lnN):
    """
    Standard Callan-Symanzik β = dα / d ln N
    Returns β aligned with N = 1 … len(alpha)-2
    """
    d_alpha = np.diff(alpha)
    d_lnN   = np.diff(lnN)
    # guard against zero / negative steps
    d_lnN   = np.where(np.abs(d_lnN) < 1e-30, np.nan, d_lnN)
    return d_alpha / d_lnN

def make_beta_alpha_weighted(alpha, lnN):
    """
    Alternative β-function: β = α · dα / d ln N
    Returns β aligned with N = 1 … len(alpha)-2
    """
    d_alpha = np.diff(alpha)
    d_lnN   = np.diff(lnN)
    # guard against zero / negative steps
    d_lnN   = np.where(np.abs(d_lnN) < 1e-30, np.nan, d_lnN)
    core    = d_alpha / d_lnN
    # Align alpha with the core derivative
    return alpha[1:len(core)+1] * core

# ------------------------------------------------------------
# 2. Single-c₃ diagnostic (default c₃ = 1)
# ------------------------------------------------------------
c3_default = 1.0
alpha = make_alpha(c3_default)

# α is defined on N = 0 … cycles
# β is defined on the intervals, we attach it to N = 1 … cycles-1
beta  = make_beta_alpha_weighted(alpha, lnN_full)
N_beta = np.arange(1, len(beta) + 1)

print(f"Initial α (Form 3, c₃={c3_default}): {alpha[0]:.8f}")
print(f"Final   α (Form 3, c₃={c3_default}): {alpha[-1]:.8f}")
# Beta is one element shorter than alpha, so use the last valid element
print(f"Initial β: {beta[0]:.6e}")
print(f"Final   β: {beta[-1]:.6e}")

# ------------------------------------------------------------
# 3. Plots — Running α and numerical β
# ------------------------------------------------------------
plt.style.use('dark_background')
fig, axes = plt.subplots(2, 1, figsize=(12, 9), sharex=True)
fig.suptitle(r'Route C, Form 3: Running Coupling and Beta Function', fontsize=16)

axes[0].plot(np.arange(len(alpha)), alpha, color='C0',
             label=r'$\alpha(A)=c_3\,w(A)[1-w(A)]$')
axes[0].set_title(r'Running $\alpha(N)$ (Route C, Form 3)')
axes[0].set_ylabel(r'$\alpha$')
axes[0].grid(True, alpha=0.3)
axes[0].legend(fontsize=9)

axes[1].plot(N_beta, beta, color='C1',
             label=r'$β = \alpha \cdot \mathrm{d}\alpha/\mathrm{d}\ln N$')
axes[1].axhline(0, color='white', ls='--', lw=0.7)
axes[1].set_title(r'Numerical $\beta$-function (Route C, Form 3)')
axes[1].set_xlabel('Cycle (N)')
axes[1].set_ylabel(r'$β$')
# Use symlog to handle potential negative values and large range
axes[1].set_yscale('symlog', linthresh=1e-12)
axes[1].grid(True, alpha=0.3)
axes[1].legend(fontsize=9)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()

# ------------------------------------------------------------
# 4. Ratio to one-loop QED target  2α²/(3π)
# ------------------------------------------------------------
# Align α with the β array (use left-edge value of each interval)
alpha_at_beta = alpha[1:len(beta)+1]
target_beta   = 2.0 * alpha_at_beta**2 / (3.0 * np.pi)

# Safe ratio
with np.errstate(divide='ignore', invalid='ignore'):
    ratio = beta / target_beta
ratio[~np.isfinite(ratio)] = np.nan

# Early-time window
early_limit = min(30, len(N_beta))
early_mask  = N_beta < early_limit
early_ratio = ratio[early_mask]

print(f"\nEarly-time ratio (N < {early_limit}):")
print(f"  Mean : {np.nanmean(early_ratio):.4f}")
print(f"  Std  : {np.nanstd(early_ratio):.4f}")

# --- Dynamic Sigma-band analysis ---
# Define sigma_for_a_band dynamically based on the initial amplitude (A_d[1] corresponds to N=1)
# This is done based on the instructions: "dynamically set the sigma-band based on the trajectory's amplitude A_d (e.g., using np.sqrt(A_d[1]))"
if len(A_d) > 1:
    sigma_for_a_band = np.sqrt(A_d[1]) # Use A_d at N=1 for setting the scale
else:
    sigma_for_a_band = sigma # Fallback to global sigma if A_d is too short

# A ~ sigma_for_a_band^2 band
A_at_beta = A_d[1:len(beta)+1] # Ensure A_d is aligned with beta

A_band_mask  = (A_at_beta >= 0.5 * sigma_for_a_band**2) & (A_at_beta <= 2.0 * sigma_for_a_band**2)

sig_ratio = ratio[A_band_mask]
if sig_ratio.size and np.isfinite(sig_ratio).any():
    print(f"\nSigma-band ratio (0.5σ² \u2264 A \u2264 2σ², where σ is based on A_d[1]):")
    print(f"  Mean : {np.nanmean(sig_ratio):.4f}")
    print(f"  Std  : {np.nanstd(sig_ratio):.4f}")
else:
    print("\nNo points inside the dynamically defined \u03C3-band for this trajectory or all ratios are non-finite.")


fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(N_beta, ratio, color='C2', label=r'$β_{\rm num}/(2\alpha^2/3\pi)$')
ax.axhline(1.0, color='red', ls='--', label='Target = 1')
if early_mask.any():
    ax.plot(N_beta[early_mask], ratio[early_mask], 'o', color='gold',
            ms=4, label=f'Early-time (N < {early_limit})')

# Plot sigma-band points if any
if sig_ratio.size and np.isfinite(sig_ratio).any():
    ax.plot(N_beta[A_band_mask], ratio[A_band_mask], 'x', color='cyan', ms=5, label=f'Sigma-band (based on A_d[1])')

ax.set_yscale('symlog', linthresh=1.0) # Use symlog for ratio as well, 1.0 as linthresh seems reasonable
ax.set_xlabel('Cycle (N)')
ax.set_ylabel('Ratio')
ax.set_title(r'Ratio of Numerical $\beta$ to QED-like Target')
ax.grid(True, which='both', alpha=0.3)
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 5. Systematic c₃ scan
# ------------------------------------------------------------
def evaluate_c3(c3_val, early_N=30):
    al = make_alpha(c3_val)
    # Use the alpha-weighted beta function
    be = make_beta_alpha_weighted(al, lnN_full)
    n  = np.arange(1, len(be)+1)
    # Align alpha with the calculated beta values
    al_b = al[1:len(be)+1]
    tgt  = 2.0 * al_b**2 / (3.0 * np.pi)
    with np.errstate(divide='ignore', invalid='ignore'):
        r = be / tgt
    r[~np.isfinite(r)] = np.nan

    em = n < early_N
    early_r = r[em]

    # Use A_d aligned with beta
    A_b = A_d[1:len(be)+1]

    # Dynamically define sigma for this evaluation run
    if len(A_d) > 1:
        local_sigma_for_a_band = np.sqrt(A_d[1]) # Use A_d at N=1 for setting the scale
    else:
        local_sigma_for_a_band = sigma # Fallback to global sigma if A_d is too short

    sm  = (A_b >= 0.5*local_sigma_for_a_band**2) & (A_b <= 2.0*local_sigma_for_a_band**2)
    sig_r = r[sm]

    return {
        'early_mean': np.nanmean(early_r) if early_r.size and np.isfinite(early_r).any() else np.nan,
        'early_std' : np.nanstd(early_r)  if early_r.size and np.isfinite(early_r).any() else np.nan,
        'sigma_mean': np.nanmean(sig_r)   if sig_r.size and np.isfinite(sig_r).any() else np.nan,
        'sigma_std' : np.nanstd(sig_r)    if sig_r.size and np.isfinite(sig_r).any() else np.nan,
    }

c3_grid = np.linspace(0.1, 10.0, 100)
early_means, early_stds = [], []
sigma_means, sigma_stds = [], [] # Also collect sigma band results

for c3 in c3_grid:
    res = evaluate_c3(c3)
    early_means.append(res['early_mean'])
    early_stds.append(res['early_std'])
    sigma_means.append(res['sigma_mean'])
    sigma_stds.append(res['sigma_std'])

early_means = np.asarray(early_means)
early_stds  = np.asarray(early_stds)
sigma_means = np.asarray(sigma_means)
sigma_stds  = np.asarray(sigma_stds)

fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)
fig.suptitle(r'Systematic $c_3$ Scan for $\beta$ Ratio', fontsize=16)

# Early-time window results
axes[0].plot(c3_grid, early_means, color='C0', label='Early-time mean ratio')
axes[0].fill_between(c3_grid,
                     early_means - early_stds,
                     early_means + early_stds,
                     color='C0', alpha=0.2, label=r'$\pm$1 std (Early-time)')
axes[0].axhline(1.0, color='red', ls='--', label='Target = 1')
axes[0].set_ylabel('Mean Ratio')
axes[0].set_title(r'Early-time window mean ratio vs $c_3$')
axes[0].grid(True, alpha=0.3)
axes[0].legend(fontsize=9)

# Sigma-band mean results
axes[1].plot(c3_grid, sigma_means, color='C2', label='Sigma-band mean ratio')
axes[1].fill_between(c3_grid,
                     sigma_means - sigma_stds,
                     sigma_means + sigma_stds,
                     color='C2', alpha=0.2, label=r'$\pm$1 std (Sigma-band)')
axes[1].axhline(1.0, color='red', ls='--', label='Target = 1')

axes[1].set_xlabel(r'$c_3$')
axes[1].set_ylabel('Mean Ratio')
axes[1].set_title(r'Dynamically defined Sigma-band mean ratio vs $c_3$')
axes[1].grid(True, alpha=0.3)
axes[1].legend(fontsize=9)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()

# Best c₃ (closest mean to 1, then smallest std)
valid_early = np.isfinite(early_means)
if valid_early.any():
    idx_mean = np.nanargmin(np.abs(early_means - 1.0))
    idx_std  = np.nanargmin(early_stds) # Smallest std, regardless of mean's proximity to 1
    print(f"\nBest c₃ by mean\u22481 (Early-time): {c3_grid[idx_mean]:.4f}  "
          f"(mean={early_means[idx_mean]:.4f}, std={early_stds[idx_mean]:.4f})")
    print(f"Best c₃ by min std (Early-time): {c3_grid[idx_std]:.4f}  "
          f"(mean={early_means[idx_std]:.4f}, std={early_stds[idx_std]:.4f})")
else:
    print("\nNo valid c₃ results for early-time window.")

valid_sigma = np.isfinite(sigma_means)
if valid_sigma.any():
    idx_mean_sigma = np.nanargmin(np.abs(sigma_means - 1.0))
    idx_std_sigma  = np.nanargmin(sigma_stds)
    print(f"\nBest c₃ by mean\u22481 (Sigma-band): {c3_grid[idx_mean_sigma]:.4f}  "
          f"(mean={sigma_means[idx_mean_sigma]:.4f}, std={sigma_stds[idx_mean_sigma]:.4f})")
    print(f"Best c₃ by min std (Sigma-band): {c3_grid[idx_std_sigma]:.4f}  "
          f"(mean={sigma_means[idx_std_sigma]:.4f}, std={sigma_stds[idx_std_sigma]:.4f})")
else:
    print("\nNo valid c₃ results for sigma-band window.")
```
