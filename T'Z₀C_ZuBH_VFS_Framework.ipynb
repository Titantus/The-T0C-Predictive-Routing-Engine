```python
# @title
# ============================================================
# 2. Discrete Solver
# ============================================================
def discrete_transition(steps, X0=1.0, Y0=0.0, Phi0=0.0, sigma=1.0):
    X   = np.empty(steps+1)
    Y   = np.empty(steps+1)
    Phi = np.empty(steps+1)

    X[0], Y[0], Phi[0] = X0, Y0, Phi0

    c, s = np.cos(theta_0), np.sin(theta_0)

    for n in range(steps):
        r2  = X[n]*X[n] + Y[n]*Y[n]
        w   = 1.0 / (1.0 + r2 / sigma**2)
        X[n+1] = (1-gamma)*w*( X[n]*c - Y[n]*s )
        Y[n+1] = (1-gamma)*w*( X[n]*s + Y[n]*c )
        Phi[n+1] = Phi[n] + gamma*r2
    return X, Y, Phi
```

```python
# @title
# ============================================================
# 3. Continuous Solver
# ============================================================
def continuous_rhs(t, S, sigma):
    X, Y, Phi = S
    r2  = X*X + Y*Y
    lam = r2 / (sigma*sigma + r2)
    dX  = -(gamma + lam)*X - theta_0*Y
    dY  =  theta_0*X - (gamma + lam)*Y
    dPhi= gamma * r2
    return [dX, dY, dPhi]
```

```python
# @title
# ============================================================
# 4. Numerical Integration
# ============================================================
cycles = 10000

# Solve the continuous system once with dense_output=True
sol = solve_ivp(continuous_rhs, [0, cycles], [1.0, 0.0, 0.0],
                args=(sigma,), rtol=1e-9, atol=1e-11, method="RK45",
                dense_output=True)
assert sol.success, sol.message

# Evaluate the solution at fine time steps for plotting
t_fine = np.linspace(0, cycles, 5*cycles+1)
S_c_fine = sol.sol(t_fine)
X_c, Y_c, Phi_c = S_c_fine[0], S_c_fine[1], S_c_fine[2]

# Evaluate the solution at integer time steps for comparison with discrete
t_int = np.arange(cycles+1)
S_c_int = sol.sol(t_int)
X_ci, Y_ci, Phi_ci = S_c_int[0], S_c_int[1], S_c_int[2]

# Solve the discrete system
X_d, Y_d, Phi_d = discrete_transition(cycles)

print(f"\nΦ_∞ discrete   = {Phi_d[-1]:.6f}")
print(f"Φ_∞ continuous = {Phi_ci[-1]:.6f}")
print(f"max |X_d − X_c| on integers = {np.max(np.abs(X_d-X_ci)):.4e}")
print(f"max |Y_d − Y_c| on integers = {np.max(np.abs(Y_d-Y_ci)):.4e}")
print(f"max |Φ_d − Φ_c| on integers = {np.max(np.abs(Phi_d-Phi_ci)):.4e}")
```

```python
# @title
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from typing import Dict, Tuple, Optional

# Optional progress bar
try:
    from tqdm import trange
except Exception:
    trange = range

# ---------------------------
# Model constants (single source)
# ---------------------------
SIN_THETA_0 = 1.0 / 3.0
P_C = 5.0 / (32.0 * np.pi)               # tensile threshold
GAMMA = 5.0 / (96.0 * np.pi**2)          # linewidth / friction
DEFAULT_R = 64
DEFAULT_STEPS = 5000
DT = 1.0
ALPHA_C = 137.75937                      # normalization used in alpha_canonical
EPS = 1e-30                              # small eps to protect divisions

# ---------------------------
# Utility functions
# ---------------------------
def alpha_canonical(Phi: np.ndarray, C: float = ALPHA_C) -> np.ndarray:
    """Return canonical α(Phi) = 1/(C*(1+Phi))."""
    return 1.0 / (C * (1.0 + Phi))


def compute_beta_and_K(alpha: np.ndarray, N: np.ndarray) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Compute beta = alpha * dα/d ln N and K = beta / target,
    where target = (2/(3π)) * α^2 is the 1-loop QED-inspired reference.

    Returns:
      (beta, target, K)
    """
    # compute derivative dα / d(ln N) robustly using gradient
    # Ensure N contains positive values to compute log
    if np.any(N <= 0):
        raise ValueError("N must contain positive values for log spacing.")
    lnN = np.log(N)
    # If alpha is constant this will return zeros appropriately
    d_alpha_d_lnn = np.gradient(alpha, lnN, edge_order=2)
    beta = alpha * d_alpha_d_lnn

    target = (2.0 / (3.0 * np.pi)) * alpha**2
    # safe division: only divide where |target| > EPS; otherwise set to np.nan
    safe_mask = np.abs(target) > EPS
    K = np.full_like(beta, np.nan, dtype=float)
    K[safe_mask] = beta[safe_mask] / target[safe_mask]
    return beta, target, K

# ---------------------------
# Simulations
# ---------------------------
def simulate_radial_slip(
    R: int = DEFAULT_R,
    steps: int = DEFAULT_STEPS,
    dt: float = DT,
    omega_core: float = SIN_THETA_0,
    p_c: float = P_C,
    gamma: float = GAMMA,
    progress: bool = False,
) -> np.ndarray:
    """
    Simple radial slip simulation that returns phi_hist (global residue over time).
    Each time step: hub advances, outer shells track or slip depending on local strain.
    """
    A = np.ones(R)
    theta = np.zeros(R)
    phi = 0.0
    phi_hist = []

    loop = trange(steps) if (progress and trange is not range) else range(steps)
    for _ in loop:
        theta[0] += omega_core * dt
        for r in range(1, R):
            strain = theta[r - 1] - theta[r]
            if strain < p_c:
                coupling = A[r - 1] * A[r] * SIN_THETA_0
                theta[r] += coupling * strain * dt
            else:
                phi += gamma * A[r]
                theta[r] += (strain - gamma) * dt
                A[r] *= (1.0 - gamma)
        phi_hist.append(phi)

    return np.asarray(phi_hist)


def simulate_force_fields(
    R: int = DEFAULT_R,
    steps: int = 4000,
    dt: float = DT,
    omega_core: float = SIN_THETA_0,
    p_c: float = P_C,
    gamma: float = GAMMA,
    progress: bool = False,
) -> Dict[str, np.ndarray]:
    """
    Simulate and record full force-field history. Returns dict with fields:
      'A', 'theta', 'strain', 'force', 'slip', 'phi', 'N'
    Shapes:
      A: (steps, R)
      theta: (steps, R)
      strain: (steps, R-1)
      force: (steps, R-1)
      slip: (steps, R-1) boolean
      phi: (steps,)
      N: (steps,)
    """
    A = np.ones(R)
    theta = np.zeros(R)
    phi = 0.0

    A_hist = np.zeros((steps, R))
    theta_hist = np.zeros((steps, R))
    strain_hist = np.zeros((steps, R - 1))
    force_hist = np.zeros((steps, R - 1))
    slip_hist = np.zeros((steps, R - 1), dtype=bool)
    phi_hist = []

    loop = trange(steps) if (progress and trange is not range) else range(steps)
    for t in loop:
        theta[0] += omega_core * dt
        for r in range(1, R):
            strain = theta[r - 1] - theta[r]
            strain_hist[t, r - 1] = strain
            if strain < p_c:
                coupling = A[r - 1] * A[r] * SIN_THETA_0
                force = coupling * strain
                theta[r] += force * dt
                slip_hist[t, r - 1] = False
            else:
                force = 0.0
                phi += gamma * A[r]
                theta[r] += (strain - gamma) * dt
                A[r] *= (1.0 - gamma)
                slip_hist[t, r - 1] = True
            force_hist[t, r - 1] = force

        A_hist[t] = A.copy()
        theta_hist[t] = theta.copy()
        phi_hist.append(phi)

    return {
        "A": A_hist,
        "theta": theta_hist,
        "strain": strain_hist,
        "force": force_hist,
        "slip": slip_hist,
        "phi": np.asarray(phi_hist),
        "N": np.arange(steps),
    }

# ---------------------------
# Plotting helpers
# ---------------------------
def plot_alpha_K(N: np.ndarray, alpha: np.ndarray, K: np.ndarray, out_path: Optional[str] = None):
    """Plot alpha(N) and K(N)."""
    fig, axes = plt.subplots(2, 1, figsize=(10, 8), sharex=True)
    axes[0].plot(N, alpha, color='C2', lw=1.5)
    axes[0].set_ylabel(r'$\alpha(N)$')
    axes[0].set_title('Running Coupling α(N)')

    axes[1].plot(N, K, color='C0', lw=1.5)
    axes[1].set_xlabel('Simulation step (N)')
    axes[1].set_ylabel(r'Dynamic $K(N)$')
    axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    if out_path:
        fig.savefig(out_path, dpi=150)
    plt.show()


def plot_force_fields(fields: Dict[str, np.ndarray], save_path: Optional[str] = None):
    """Multi-panel force-field plotting similar to original script."""
    steps = fields['N'].size
    R = fields['A'].shape[1]

    plt.style.use('dark_background')
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    # A(r,t)
    ax = axes[0, 0]
    im = ax.imshow(fields['A'].T, aspect='auto', origin='lower', extent=[0, steps, 0, R], cmap='magma')
    ax.set_title(r'Amplitude Field $A(r,t)$')
    ax.set_xlabel('Step $N$')
    ax.set_ylabel('Radial shell $r$')
    plt.colorbar(im, ax=ax, label='$A$')

    # strain (log)
    ax = axes[0, 1]
    strain_pos = np.clip(fields['strain'], 1e-12, None)
    im = ax.imshow(strain_pos.T, aspect='auto', origin='lower', extent=[0, steps, 0, R-1], cmap='viridis', norm=LogNorm())
    ax.set_title(r'Strain Field $|\theta_{r-1}-\theta_r|$ (log)')
    ax.set_xlabel('Step $N$')
    ax.set_ylabel('Interface $r$')
    plt.colorbar(im, ax=ax, label='strain')

    # force
    ax = axes[1, 0]
    im = ax.imshow(fields['force'].T, aspect='auto', origin='lower', extent=[0, steps, 0, R-1], cmap='coolwarm')
    ax.set_title(r'Local Coupling Force $F(r,t)$')
    ax.set_xlabel('Step $N$')
    ax.set_ylabel('Interface $r$')
    plt.colorbar(im, ax=ax, label='force')

    # slip events + phi overlay
    ax = axes[1, 1]
    ax.imshow(fields['slip'].T, aspect='auto', origin='lower', extent=[0, steps, 0, R-1], cmap='binary', alpha=0.7)
    ax.set_title('Slip Events (white = yielded)')
    ax.set_xlabel('Step $N$')
    ax.set_ylabel('Interface $r$')

    ax2 = ax.twinx()
    ax2.plot(fields['N'], fields['phi'], color='C1', lw=2, label=r'$\Phi(N)$')
    ax2.set_ylabel(r'Global Residue $\Phi$', color='C1')
    ax2.tick_params(axis='y', labelcolor='C1')
    ax2.legend(loc='upper left')

    plt.tight_layout()
    if save_path:
        fig.savefig(save_path, dpi=150)
    plt.show()

# ---------------------------
# Example main flow
# ---------------------------
def main():
    # Quick run parameters
    R = DEFAULT_R
    steps = DEFAULT_STEPS
    dt = DT

    # THE FIX: Calculate the elastic transmission limit
    gamma_trans = P_C * SIN_THETA_0

    # Drive the core just 5% over the limit to induce a stable, rolling slip
    drive_omega = gamma_trans * 1.05

    print("Simulating radial slip-pivot (compact)...")
    # Pass omega_core=drive_omega into the simulation
    phi_hist = simulate_radial_slip(R=R, steps=steps, dt=dt, omega_core=drive_omega, progress=False)
    N = np.arange(1, phi_hist.size + 1)

    print(f"  Final Φ = {phi_hist[-1]:.6e}")

    alpha = alpha_canonical(phi_hist)
    beta, target, K = compute_beta_and_K(alpha, N)

    # Diagnostics for K (skip NaNs)
    finite_inds = np.isfinite(K)
    if finite_inds.any():
        finite_idx = np.nonzero(finite_inds)[0]
        print(f"  K length (finite): {finite_inds.sum()}")
        print(f"  K initial (first finite): {K[finite_idx[0]]:.6g}")
        print(f"  K final (last finite): {K[finite_idx[-1]]:.6g}")
    else:
        print("  K is NaN everywhere (target too small or α constant)")

    # Optionally run full force-field simulation for visual diagnostics
    print("Running force-field simulation (detailed) ...")
    # Pass omega_core=drive_omega here as well
    fields = simulate_force_fields(R=R, steps=min(4000, steps), dt=dt, omega_core=drive_omega, progress=False)
    print(f"  Final global residue Φ = {fields['phi'][-1]:.6f}")

    # Plotting
    plot_force_fields(fields)
    plot_alpha_K(N, alpha, K)


if __name__ == "__main__":
    main()
```
